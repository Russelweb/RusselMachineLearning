```python
from flask import Flask, request, jsonify
from gtts import gTTS
import os
from playsound import playsound

app = Flask(__name__)

@app.route('/speak', methods=['POST'])
def speak():
    # Get the JSON data from the request
    data = request.json
    text = data.get('text', '')

    if not text:
        return jsonify({'error': 'No text provided'}), 400
    
    # Create the text-to-speech object
    audio_file = 'output.mp3'
    text_to_speech = gTTS(text=text, lang='en', slow=False)

    # Save the audio file
    text_to_speech.save(audio_file)

    # Play the audio file (optional, you can remove this if you only want to save)
    playsound(audio_file)

    return jsonify({'message': 'Speech generated', 'audio_file': audio_file}), 200

if __name__ == '__main__':
    app.run(debug=True)
```
