```python
import pyttsx3
import os

text = 'This is a trial speech to test my new model which is a text-to-speech model which takes in a text as input and return the text in speech.'

# Initialize the text-to-speech engine
text_to_speech = pyttsx3.init()

# Adjust the speech rate
text_to_speech.setProperty('rate', 150)

# Get available voices and allow user selection
voices = text_to_speech.getProperty('voices')
for index, voice in enumerate(voices):
    print(f"Voice {index}: {voice.name}")

# Set voice (0 for male, 1 for female)
voice_index = int(input("Select voice (0 or 1): "))
text_to_speech.setProperty('voice', voices[voice_index].id)

# Speak the text
text_to_speech.say(text)

# Save the audio to a file
text_to_speech.save_to_file(text, 'test.wav')  # Save as WAV for better compatibility

# Run and wait1

text_to_speech.runAndWait()
```
